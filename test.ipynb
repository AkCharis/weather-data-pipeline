```python
import pandas as pd
import requests
import sqlalchemy
import time

# A list of cities to get forecast data
cities = [{'city': 'Lagos', 'latitude': 6.5244, 'longitude': 3.3792},
    {'city': 'London', 'latitude': 51.5074, 'longitude': -0.1278},
    {'city': 'New York', 'latitude': 40.7128, 'longitude': -74.0060},
    {'city': 'Ibadan', 'latitude': 7.3776, 'longitude': 3.9059},]

city_data_list = []

for city in cities:
    params = {
        'latitude': city['latitude'],
        'longitude': city['longitude'],
        'hourly': ['temperature_2m' ,'precipitation', 'wind_speed_10m']
    }
    
    response = requests.get('https://api.open-meteo.com/v1/forecast', params=params, timeout=30)
    time.sleep(3)

    raw_data = response.json()

    data = pd.DataFrame(raw_data['hourly'])
    data['city_name'] = city['city']
    data['extracted_at_utc'] = pd.Timestamp.now(tz='UTC')

    city_data_list.append(data)

    

df = pd.concat([x for x in city_data_list])

df

"""
params = {
    'latitude': 6.5244,
    'longitude': 3.3792,
    'hourly': ['temperature_2m' ,'precipitation', 'wind_speed_10m']
}

response = requests.get('https://api.open-meteo.com/v1/forecast', params=params)
"""




```
